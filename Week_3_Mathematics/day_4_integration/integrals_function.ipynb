```python
import sympy as sp
```

```python
def def_integral(function, x, a, b):
    return sp.integrate(function, (x, a, b))

def indef_integral(function, x):
    return sp.integrate(function, x)
```

```python
x = sp.Symbol("x")
x
```

```python
print(def_integral(x**2, x, 0, 3))
print(indef_integral(x**3 + 2*x, x))
```
