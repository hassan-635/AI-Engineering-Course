```python
import sympy as sp
```

```python
w, x, y = sp.symbols("w x y")
```

```python
prediction = w*x
loss = (y - prediction)**2
```

```python
derivative = sp.diff(loss, w)
print(f"{prediction} => {loss} => {derivative}")
```

```python
data = []
for i in range(1, 101):
    data.append((i, 3*i))
```

```python
data
```

```python
weight = 1
learning_rate = 0.01
beta1 = 0.9
beta2 = 0.999
m=0
v=0
t=0
epsilon = 1e-8
```

```python
for epoch in range(5):
    print(f"Epoch => {epoch+1} :")
    for x_value, y_value in data:
        gradient_value = derivative.subs({
            x:x_value,
            y:y_value,
            w:weight
        })
        print(gradient_value)
        gradient = float(gradient_value)

        t+=1

        m = beta1*m+(1-beta1)*gradient
        v = beta2*v+(1-beta2)*(gradient**2)

        m_hat = m/(1-beta1**t)
        v_hat = v/(1-beta2**t)

        weight = weight - learning_rate * (m_hat/(v_hat**0.5+epsilon))

    print("Weight =", round(weight,4))
print("\nFinal Weight =", round(weight,4))
```
