```python
import sympy as sp
```

```python
learning_rate = 0.1
weight = 1
data = [
    (1, 2),
    (2, 4),
    (3, 6)
]
```

```python
w, x, y = sp.symbols("w x y")
```

```python
prediction = w*x
```

```python
loss = (y - prediction)**2
```

```python
derivative = sp.diff(loss, w)
derivative
```

```python
for epoch in range(5):
    print(f"Ëpoch : {epoch+1}")
    for x_value, y_value in data:

        prediction_value = prediction.subs({
            w:weight,
            x:x_value
        })

        loss_value = loss.subs({
            w:weight,
            x:x_value,
            y:y_value
        })

        gradient = derivative.subs({
            w:weight,
            x:x_value,
            y:y_value
        })

        weight = weight - learning_rate * gradient

        print(f"Prediction -> {prediction_value}")
        print(f"Loss -> {loss_value}")
        print(f"Gradient -> {gradient}")
        print(f"New Weight : {weight}")
        print("___________________________________")
```
