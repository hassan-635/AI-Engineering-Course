```python
import sympy as sp
```

```python
w, x, y = sp.symbols("w x y")
prediction = w*x
loss = (y - prediction)**2
```

```python
print(prediction, loss)
```

```python
learning_rate = 0.001
data = []
weight = 1
batch_size = 32
```

```python
for i in range(1, 65):
    data.append((i, i*2))

data
```

```python
derivative = sp.diff(loss, w)
```

```python
derivative
```

```python
for epoch in range(5):
    print(f"Epoch : => {epoch+1}")
    for i in range(0, len(data), batch_size):
        batch = data[i:i+batch_size]
        total_gradient = 0
        for x_value, y_value in batch:
            gradient = derivative.subs({
                x:x_value,
                y:y_value,
                w:weight
            })
            total_gradient += gradient
        avg_gradient = total_gradient/len(batch)
        weight = weight - learning_rate * gradient
        print(f"Batch {(i//batch_size)+1}")
        print("Average Gradient =", round(float(avg_gradient),4))
        print("Updated Weight =", round(float(weight),4))
        print("--------------------------------")
print("\nFinal Weight =", round(float(weight),4))
            
```
