```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

```python
x = sp.Symbol("x")
```

```python
x_points = []
y_points = []
losses = []
```

```python
x_value = 8
learning_rate = 0.1
```

```python
f = x**2
```

```python
df = sp.diff(f, x)
```

```python
for i in range(15):
    y_value = f.subs(x, x_value)
    x_points.append(float(x_value))
    y_points.append(float(y_value))
    losses.append(float(y_value))
    gradient = df.subs(x, x_value)
    x_value = x_value - learning_rate * gradient
```

```python
plt.plot(range(1, 16), losses, marker='o')
plt.title("Loss vs Iterations")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```
