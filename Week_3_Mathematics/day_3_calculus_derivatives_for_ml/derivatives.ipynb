# Derivative 

```python
import sympy as sp
```

```python
x = sp.Symbol("x")
f = x**2
derivative = sp.diff(f)
print(derivative)
```

# Partial Derivatives

```python
x, y = sp.symbols("x y")
f = x**2 + y**2
grad_x = sp.diff(f, x)
grad_y = sp.diff(f, y)
print(f"Partial Derivaties :  {grad_x} {grad_y}")
```

# Function to find derivatives and partial dewrivatives

```python
def derivative(function, x):
    return sp.diff(function, x)

def partial_derivative(function, x, y):
    grad_x = sp.diff(function, x)
    grad_y = sp.diff(function, y)
    tup = (0, 0)
    tup[0] = grad_x
    tup[1] = grad_y
    return tup
```

```python
x = sp.Symbol("x")
print(derivative(x**2, x))
print(derivative(x**3+3*(x**2), x))
```
