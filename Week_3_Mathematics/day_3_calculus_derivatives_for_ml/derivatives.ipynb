# Derivative 

```python
import sympy as sp
```

```python
x = sp.Symbol("x")
f = x**2
derivative = sp.diff(f)
print(derivative)
```

# Partial Derivatives

```python
x, y = sp.symbols("x y")
f = x**2 + y**2
grad_x = sp.diff(f, x)
grad_y = sp.diff(f, y)
print(f"Partial Derivaties :  {grad_x} {grad_y}")
```
