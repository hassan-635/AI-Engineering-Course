w1 = 5
w2 = 10
x1 = 6
x2 =3

```python
import sympy as sp
```

```python
w1, w2 = sp.symbols("w1 w2")
x1, x2 = sp.symbols("x1 x2")
print(w1, w2, x1, x2)
```

```python
prediction = w1*x1 + w2*x2
print(prediction)
```

```python
prediction_value = prediction.subs({
    w1:5,
    w2:3,
    x1:6,
    x2:3
})

prediction_value
```
