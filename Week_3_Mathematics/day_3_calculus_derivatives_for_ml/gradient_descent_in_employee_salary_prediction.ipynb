w1 = 5
w2 = 10
x1 = 6
x2 =3

```python
import sympy as sp
```

```python
w1, w2 = sp.symbols("w1 w2")
x1, x2 = sp.symbols("x1 x2")
print(w1, w2, x1, x2)
```

```python
prediction = w1*x1 + w2*x2
print(prediction)
```

```python
prediction_value = prediction.subs({
    w1:5,
    w2:3,
    x1:6,
    x2:3
})

prediction_value
```

```python
y = 100
```

```python
loss = (y - prediction)**2
```

```python
loss_value = loss.subs({
    w1:5,
    w2:3,
    x1:6,
    x2:3,
    y:100
})

loss_value
```

```python
grad_w1 = sp.diff(loss, w1)
grad_w2 = sp.diff(loss, w2)
print("Gradients : ")
print(grad_w1)
print(grad_w2)
```

```python
grad_w1_value = grad_w1.subs({
    w1:5,
    w2:3,
    x1:6,
    x2:3
})

grad_w2_value = grad_w2.subs({
    w1:5,
    w2:3,
    x1:6,
    x2:3
})

print("Gradients Values : ")
print(grad_w1_value)
print(grad_w2_value)
```

```python
learning_rate = 0.01
losses = []
x1_value = 6
x2_value=3
y_value = 100
weight1 = 5
weight2 = 10
```

```python
for epoch in range(10):
    prediction_value2 = prediction.subs({
        w1: weight1,
        w2: weight2,
        x1: x1_value,
        x2: x2_value,
        y: y_value
    })

    loss_value2 = loss.subs({
        w1: weight1,
        w2: weight2,
        x1: x1_value,
        x2: x2_value,
        y: y_value
    })

    losses.append(float(loss_value2))

    grad_w1_value2 = grad_w1.subs({
        w1: weight1,
        w2: weight2,
        x1: x1_value,
        x2: x2_value,
        y: y_value
    })

    grad_w2_value2 = grad_w2.subs({
        w1: weight1,
        w2: weight2,
        x1: x1_value,
        x2: x2_value,
        y: y_value
    })

    weight1 = weight1 - learning_rate * grad_w1_value2
    weight2 = weight2 - learning_rate * grad_w2_value2

    print(f"Epoch : {epoch+1} -> Prediction : {prediction_value2} -> Loss : {loss_value2}")
    print(f"Gradients : {grad_w1_value2} & {grad_w2_value2}")
    print(f"New Weights : {weight1} & {weight2}")
    print("_______________________________________________________________________________")
```

```python
losses
```

```python
import matplotlib.pyplot as plt

iterations = range(1, len(losses)+1)

plt.plot(iterations, losses, marker="o")

plt.title("Loss vs Iterations")
plt.xlabel("Iterations")
plt.ylabel("Loss")

plt.grid(True)

plt.show()
```
