```python
x = [1, 2, 3]
y = [2, 4, 6]
```

```python
initial_weight = 0
```

```python
learning_rate = 0.1
```

```python
for epoch in range(10):
    for i in range(len(x)):
        prediction = initial_weight*x[i]
        error = prediction - y[i]
        derivative = 2 * error * x[i]
        initial_weight = initial_weight - learning_rate * derivative
    print("Epoch : ", epoch+1, "Weight : ", round(initial_weight, 4))
    
```

```python
print("\nFinal Weight =", round(initial_weight, 4))
```
