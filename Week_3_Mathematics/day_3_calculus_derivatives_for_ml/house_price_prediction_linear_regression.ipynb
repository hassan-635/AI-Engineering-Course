# House Price Prediction using Linear Regression

```python
import sympy as sp
```

```python
w1, w2 = sp.symbols("w1 w2")
x1, x2 = sp.symbols("x1 x2")
```

```python
prediction = w1*x1 + w2*x2
```

```python
print("Prediction Formula : ", prediction)
```

```python
prediction_value = prediction.subs({
    w1:2,
    w2:1,
    x1:3,
    x2:3
})
```

```python
print(prediction_value)
```

# Actual Price is 15 but Predicted Price 9
# Now Loss Function

```python
y = sp.Symbol("y")
```

```python
loss = (y - prediction)**2
```

```python
print(loss)
```

```python
loss_value = loss.subs({
    y:15,
    w1:2,
    w2:1,
    x1:3,
    x2:3
})
```

```python
print(loss_value)
```

# Now AI Knows loss value
# Now it comes that how AI knows :
# increase or decrease w1 or w2 both to decrease loss value

```python
grad_w1 = sp.diff(loss, w1)
grad_w2 = sp.diff(loss, w2)
```

```python
print("∂L/∂w1 =", grad_w1)
print("∂L/∂w2 =", grad_w2)
```

```python
grad_w1_value = grad_w1.subs({
    w1:2,
    w2:1,
    x1:3,
    x2:3,
    y:15
})

grad_w2_value = grad_w2.subs({
    w1:2,
    w2:1,
    x1:3,
    x2:3,
    y:15
})
```

```python
print("Gradient w1 =", grad_w1_value)
print("Gradient w2 =", grad_w2_value)
```

# Now AI know which direction to move to minimize loss
# Now gradient descent will tell new weights for minimum loss

```python
learning_rate = 0.01
```

```python
new_w1 = 2 - learning_rate * grad_w1_value
new_w2 = 1 - learning_rate * grad_w2_value
```

```python
print("New w1 =", new_w1)
print("New w2 =", new_w2)
```

```python
new_prediction = prediction.subs({
    w1:new_w1,
    w2:new_w2,
    x1:3,
    x2:3
})
```

```python
print(new_prediction)
```

```python
loss_value2 = loss.subs({
    y:15,
    w1:new_w1,
    w2:new_w2,
    x1:3,
    x2:3
})
print(loss_value2)
```

# Now Apply small learning rate :

```python
w1, w2 = sp.symbols("w1 w2")
x1, x2 = sp.symbols("x1 x2")
y = sp.Symbol("y")

# Prediction Formula
prediction = w1*x1 + w2*x2

# Loss Function
loss = (y - prediction)**2

# Partial Derivatives
grad_w1 = sp.diff(loss, w1)
grad_w2 = sp.diff(loss, w2)

# Initial Weights
weight1 = 2
weight2 = 1

# Dataset
x1_value = 3
x2_value = 3
y_value = 15

# Learning Rate
learning_rate = 0.01

for i in range(10):

    # Prediction
    prediction_value = prediction.subs({
        w1: weight1,
        w2: weight2,
        x1: x1_value,
        x2: x2_value
    })

    # Loss
    loss_value = loss.subs({
        w1: weight1,
        w2: weight2,
        x1: x1_value,
        x2: x2_value,
        y: y_value
    })

    # Gradients
    grad1 = grad_w1.subs({
        w1: weight1,
        w2: weight2,
        x1: x1_value,
        x2: x2_value,
        y: y_value
    })

    grad2 = grad_w2.subs({
        w1: weight1,
        w2: weight2,
        x1: x1_value,
        x2: x2_value,
        y: y_value
    })

    # Update Weights
    weight1 = weight1 - learning_rate * grad1
    weight2 = weight2 - learning_rate * grad2

    print(f"\nIteration {i+1}")
    print("Prediction =", prediction_value)
    print("Loss =", loss_value)
    print("Gradient w1 =", grad1)
    print("Gradient w2 =", grad2)
    print("New w1 =", weight1)
    print("New w2 =", weight2)
```
