# House Price Prediction using Linear Regression

```python
import sympy as sp
```

```python
w1, w2 = sp.symbols("w1 w2")
x1, x2 = sp.symbols("x1 x2")
```

```python
prediction = w1*x1 + w2*x2
```

```python
print("Prediction Formula : ", prediction)
```

```python
prediction = prediction.subs({
    w1:2,
    w2:1,
    x1:3,
    x2:3
})
```

```python
print(prediction)
```
