# House Price Prediction using Linear Regression

```python
import sympy as sp
```

```python
w1, w2 = sp.symbols("w1 w2")
x1, x2 = sp.symbols("x1 x2")
```

```python
prediction = w1*x1 + w2*x2
```

```python
print("Prediction Formula : ", prediction)
```

```python
prediction = prediction.subs({
    w1:2,
    w2:1,
    x1:3,
    x2:3
})
```

```python
print(prediction)
```

# Actual Price is 15 but Predicted Price 9
# Now Loss Function

```python
y = sp.Symbol("y")
```

```python
loss = (y - prediction)**2
```

```python
print(loss)
```

```python
loss_value = loss.subs({
    y:15,
    w1:2,
    w2:1,
    x1:3,
    x2:3
})
```

```python
print(loss_value)
```
