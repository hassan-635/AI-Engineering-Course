```python
import sympy as sp
```

```python
def derivative(function, x, order):
    return sp.diff(function, x, order)

def partial_derivative(function, x, y):
    grad_x = sp.diff(function, x)
    grad_y = sp.diff(function, y)
    return [sp.diff(grad_x, x), sp.diff(grad_y, y), sp.diff(grad_x, y), sp.diff(grad_y, x)]
```

```python
x, y = sp.symbols("x y")
print(derivative(x**3, x, 2))
print(partial_derivative(x**2+4*(x*y)+y**2, x, y))
```
