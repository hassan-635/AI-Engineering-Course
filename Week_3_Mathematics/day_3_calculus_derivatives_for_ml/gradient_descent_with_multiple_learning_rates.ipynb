```python
import numpy as np
```

```python
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 6, 8, 10], dtype=float)
```

```python
learning_rate = [0.001, 0.01, 0.1]
```

```python
for lr in learning_rate:
    weight = 0
    n = len(x)
    print(f"Learning Rate = {lr}")
    for epoch in range(20):
        prediction = weight * x
        error = prediction - y
        derivative = (2/n) * np.sum(error * x)
        weight = weight - lr * derivative
        loss = np.mean(error**2)
        print(f"Epoch : {epoch+1:2d} -> Loss : {loss:.4f} -> Weight : {weight:.4f}")
```
