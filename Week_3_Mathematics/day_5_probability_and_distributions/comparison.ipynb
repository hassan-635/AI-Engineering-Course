```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, uniform
mean = 50
std = 10
x1 = np.linspace(10, 90, 500)
gaussian = norm.pdf(x1, mean, std)
x2 = np.linspace(10, 90, 500)
uniform_dist = uniform.pdf(x2, loc=10, scale=80)
```

```python
plt.figure(figsize=(8,5))
plt.plot(x1, gaussian)
plt.title("Gaussian Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(x2, uniform_dist)
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```
