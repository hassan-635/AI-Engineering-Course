```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multinomial
n = 10
p = [0.5, 0.3, 0.2]
outcome = [5, 3, 2]
probability = multinomial.pmf(outcome, n=n, p=p)
print("Probability:", probability)
classes = ["Cat", "Dog", "Bird"]
plt.bar(classes, outcome)
plt.title("Multinomial Distribution")
plt.xlabel("Classes")
plt.ylabel("Count")
plt.show()
```
