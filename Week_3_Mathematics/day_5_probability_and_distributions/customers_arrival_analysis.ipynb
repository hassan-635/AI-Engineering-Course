# Customer Arrivals Analysis Using Probability Distributions

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
customers = 500
```

# Dataset

```python
age = np.random.normal(35, 8, customers) # mean, std, size
waiting_time = np.random.exponential(5, customers) # scale, size means avg waiting time is approx. 5
customers_per_hour = np.random.poisson(12, customers) # lam, size means average customers per hour are 12
membership = np.random.binomial(1, 0.6, customers) # n, p, size 
```

```python
df = pd.DataFrame({
    "Age": age,
    "Waiting_Time": waiting_time,
    "Customers_Per_Hour": customers_per_hour,
    "Membership": membership
})
```
