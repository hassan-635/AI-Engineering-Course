# Matrix Decomposition Methods (One Complete AI Case Study)

# LU

```python
import numpy as np
from scipy.linalg import lu
```

```python
A = np.array([
    [8, 2, 1],
    [2, 6, 1],
    [1, 1, 5]
], dtype=float)
```

```python
print(A)
```

```python
P, lower, upper = lu(A)
```

```python
print(P)
```

```python
lower
```

```python
upper
```

```python
orig = P @ lower @ upper
```

```python
orig
```

```python
np.allclose(A, orig)
```

# QR Decomposition 

```python
A = np.array([
    [2, 1],
    [3, 2],
    [4, 3],
    [5, 4]
], dtype=float)
```

```python
A
```

```python
Q, R = np.linalg.qr(A)
```

```python
Q
```

```python
R
```

```python
Q @ R
```

```python
np.allclose(A, Q @ R)
```

# Eigen Value Decomposition

```python
A = np.array([
    [4, 2, 1],
    [2, 3, 1],
    [1, 1, 2]
], dtype=float)
```

```python
A
```

```python
eigenVals, eigenVec = np.linalg.eig(A) 
```

```python
eigenVals
```

```python
eigenVec
```

```python
lambda1 = np.diag(eigenVals)
lambda1
```

```python
orig = eigenVec @ lambda1 @ np.linalg.inv(eigenVec)
orig
```

```python
np.allclose(orig, A)
```

# Cholesky Decomposition :

```python
A = np.array([
    [6, 2, 1],
    [2, 5, 2],
    [1, 2, 4]
], dtype=float)
```

```python
A
```

```python
L = np.linalg.cholesky(A)
```

```python
L
```

```python
LT = L.T
LT
```

```python
orig = L @ LT
orig
```

```python
np.allclose(A, orig)
```
