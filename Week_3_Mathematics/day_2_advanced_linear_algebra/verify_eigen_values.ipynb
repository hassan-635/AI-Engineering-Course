# Verify the Property det(A−λI)=0

```python
import numpy as np

A = np.array([
    [4,2,1],
    [2,3,1],
    [1,1,2]
])
```

```python
eigVals, eigVec = np.linalg.eig(A)
print(eigVals)
print(eigVec)
```

```python
i = np.eye(3)
print(i)
```

```python
lambda1 = eigVals[0]
print(lambda1)
```

```python
mul = lambda1 * i
print(mul)
```

```python
sub = A - mul
print(sub)
```

```python
det = np.linalg.det(sub)
print(det)
```

```python
print(np.isclose(det, 0))
```

## So property satisfied
