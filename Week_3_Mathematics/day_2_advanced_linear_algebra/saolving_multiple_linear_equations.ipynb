# Solving Multiple Linear Equations (Determinant + Inverse)

```python
import numpy as np
```

```python
A = np.array([
    [2, 1, 3],
    [1, 2, 1],
    [3, 1, 2]
])

A
```

```python
det = np.linalg.det(A)
det
```

```python
inv = np.linalg.inv(A)
inv
```

## this is used in least squeares and linear regression
