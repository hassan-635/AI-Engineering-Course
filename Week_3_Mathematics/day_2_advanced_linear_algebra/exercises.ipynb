# Determinant and Inverse 

```python
import numpy as np
```

```python
matrix = np.array([[1, 2], [3, 4]])
print(matrix)
print(type(matrix))
```

```python
det = np.linalg.det(matrix)
det
```

```python
inverse = np.linalg.inv(matrix)
print(inverse)
```

# Eigen Values and Eigen Vectors

```python
mat = np.array([[1, 2], [3, 4]])
print(mat)
```

```python
eigvals, eigvector= np.linalg.eig(mat)
print(eigvals)
print(eigvector)
```
