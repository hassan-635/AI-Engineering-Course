# Verify the Property Av=λv

```python
import numpy as np

A = np.array([
    [4,2,1],
    [2,3,1],
    [1,1,2]
])
```

```python
eigVal, eigVec = np.linalg.eig(A)
```

```python
print(eigVal)
print(eigVec)
```

```python
lamda1 = eigVal[0]
print(lamda1)
```

```python
v = eigVec[:,0]
print(v)
```

```python
left = A @ v
right = lamda1 * v
```

```python
print(left)
print(right)
```

```python
print(np.allclose(left, right))
```
